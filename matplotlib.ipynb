```python
import matplotlib.pyplot as plt
import numpy as np
```

Basic Line Plot

```python
x = np.array([1,2,3,4])
y = np.array([10,20,25,30])

plt.plot(x,y)
plt.title("Basic line PLot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

Line Style and Colors

```python
from matplotlib.lines import lineStyles
x = np.array([1,2,3,4])
y = np.array([3,8,1,10])

plt.plot(y, linestyle='dotted', color="pink", marker='o')
plt.title("Styled Line Example")
plt.show()
```

Multiple Line in one Graphql with Numpy and Matplotlib

```python
x = np.array([1,2,3,4])
y =np.array([2,4,6,8])
y1 = np.array([1,3,5,7])

plt.plot(x, y, label='Line 1', color='red', marker='s')
plt.plot(x, y1, label='line 2',color='blue', linestyle='--', marker='o')
plt.title('Multiple Line Example')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.show()
```

Bar Chart with NUmpy and Maplotlib Example

```python

x = np.array(['A', 'B', 'C', 'D'])
y = np.array([5, 9, 1, 7])

plt.bar(x, y, color='orange')
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

Scatter Plot With Numpy and Matplotlib

```python
x = np.array([1,2,3,4,5])
y = np.array([5,4,6,3,7])

plt.scatter(x,y, color='purple', s=100)
plt.title("Scatter Plot Example")
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

Historam With Numpy and Matplotlip

```python
data = np.array([3,4,5,6,7,8,9,10])

plt.hist(data,bins=5,color='skyblue',edgecolor='black')
plt.title('Histogram Example')
plt.xlabel('Values')
plt.ylabel('Count')
plt.show()
```

Subplots(Multiple Graphs Together)

```python
x = np.array([1,2,3,4])
y1 = np.array([2,4,6,8])
y2 = np.array([3,6,9,12])

plt.figure(figsize=(6,2))

plt.subplot(1,2,1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1,2,2)
plt.bar(x, y2, color='blue')
plt.title("Plot 2")

plt.tight_layout()
plt.show()
```
